## Basics ##

__PsychRoom Tool Kit__ has built-in unit handling support in order to:

* avoid unit ambiguity,
* make functions "_unit aware_",
* and make unit conversions quick and painless.

To do this, the excellent [_Pint_](http://pint.readthedocs.org/en/latest/ "Pint: a Python units library") package has been incorporated into the the tool kit.  This tutorial aims to introduce you to the unit-handling functionality and show you how to harness its power.

###Uncertainty Calculation###

__Importing information__

When the data file was read by the io.read() function, uncertainty information in the data file has already been passed to the dictionary *metadata* under pandas Dataframe *data*. Let's try to examine the uncertainty information below.


```python
data.metadata
```

The uncertainty information is stored in a dictionary with the key name *uncertainty*. We can take a closer look of the information related to the measurement *idhx_air_in_RH* by running the following code.

```python
data.metadata['uncertainty']['idhx_air_in_RH']
```

The information is stored in a tuple with names *var*, *eqn* and *ori_unit*. To know which measurements the uncertainty of *idhx_air_in_RH* depend on, we can run the following code.

```python
data.metadata['uncertainty']['idhx_air_in_RH'].var
```

The uncertainty of *idhx_air_in_RH* is dependent on two variables: the measurement itself and *idhx_air_in_T*. To examine how their values change the uncertainty, we can look at the equation by running the following.

```python
data.metadata['uncertainty']['idhx_air_in_RH'].eqn
```

You can see that the equation is a piecewise function with two variables *x0* and *x1*. *x0* stands for the measurement itself (i.e. *idhx_air_in_RH*) and *x1* is the second variable in the list - *idhx_air_in_T*.

The information also stores the engineering units of the variables inside the equation, and you can see it with the following code.

```python
data.metadata['uncertainty']['idhx_air_in_RH'].ori_unit
```

You can see that the first variable *idhx_air_in_RH* is dimensionless in the equation, and the second variable *idhx_air_in_T* is in degC in the equation.

The equation is written in the syntax of *sympy* package and please refer to http://sympy.org/en/index.html for further questions on the equation. Notice that the units of the measurement in the equations follow the units of the measurement in your raw data file. For instance, the above equation is only correct if *idhx_air_in_RH* is dimensionless and *idhx_air_in_T* is in degree C.

### Unit Conversion ###

Unit conversion is a common occurence in any scientific application and perhaps no better example of this is code pertaining to HVAC.  Simple conversions are accomplished easily.  Below, we will convert a temperature measurement from degrees Celsius to other temperature units.

```python
print('Orignal Value:\t', data['idhx_air_in_T'].mean(), 'degC')
print('New Value:\t', data.convert('idhx_air_in_T', 'degF').mean(), 'degF')
print('New Value:\t', data.convert('idhx_air_in_T', 'kelvin').mean(), 'kelvin')
print('New Value:\t', data.convert('idhx_air_in_T', 'rankine').mean(), 'rankine')
```

What happens when we try to convert a quantity to a noncompatible unit?

```python
print('Orignal Value:\t', data['comp_ref_out_pg'].mean(), 'kPa')
print('New Value:\t', data.convert('comp_ref_out_pg', 'kW').mean(), 'kW')
```

Woah! What just happen?  Let's look at the the last line of the interpreter output:

    DimensionalityError: 
    Cannot convert from 'kilopascal' ([mass] / [length] / [time] ** 2) to 'kilowatt' ([length] ** 2 * [mass] / [time] ** 3)
    
It says that it cannot convert from 'kilopascal' to 'kilowatt' since they have different dimensionalities.  Its also clear from the error the _Pint_ library does a excellent job of handling derived units.  Knowing this, we can even convert to some obscure derived unit if we wanted, say:

```python
print('Orignal Value:\t', data['comp_ref_out_pg'].mean(), 'kPa')
print('New Value:\t', data.convert('comp_ref_out_pg', 
                                   'pennyweight / barleycorn / minute ** 2').mean(),
      'pennyweight per barleycorn-minute**2')
```

Just what you've always wanted to know.

### Important Note on How _PsychRoom TK_ handles Conversions ###

The philosophy used throughout the _PsychRoom_ codebase is to preserve the original data whenever possible.  This means that when a unit conversion is performed, the default behavior performed by _PsychRoom_ is to keep the data in the original unit and return a copy of the data coverted to the new unit.

We can see this by comparing the memory locations of the data frame before and after conversion.

```python
print("Data before conversion in [kW]:\n", data.comp_elec_in_pwr[:5],"\n")
print("New data after conversion in ['Btu/hr']:\n", data.convert('comp_elec_in_pwr', 'Btu/hr')[:5],"\n")
print("Orignal data after conversion in [kW]:\n", data.comp_elec_in_pwr[:5],"\n")
```

If we would rather _overwrite_ the converted quantities, we simply pass an optional argument to the _convert_ function.

```python
print("Data before conversion in [kW]:\n", data.comp_elec_in_pwr[:5],"\n")
print("New data after conversion in ['Btu/hr']:\n", data.convert('comp_elec_in_pwr', 'Btu/hr', overwrite=True)[:5],"\n")
print("Orignal data after conversion in [Btu/hr]:\n", data.comp_elec_in_pwr[:5],"\n")
```

Just to make sure, let's take a closer look at the unit of the compressor electric inlet power column.

```python
print("New Unit: ", data.units('comp_elec_in_pwr'))
```

### Diving Deeper ###

Converting a particular quantity from one unit to another is useful, but it might be more useful to convert all measurements of one dimension to a different or standard unit.  Fortunately, we can do this simply as follows:

```python
data.convert('[pressure]', 'psi')[:5]
```

Again, we can overwrite values using the same optional parameter as before.

```python
data.convert('[power]', 'W', overwrite=True)[:5]
```

We can also do a unit conversion with the uncertainty. For instance, we can generate a dataframe consisting of *idhx_air_in_T* and its uncertainty in degF.

```python
idhx_air_in_T_deg_F = data.convert('idhx_air_in_T', 'degF', cal_uncer=True)
print("Temperature values:")
print(idhx_air_in_T_deg_F)
print("Uncertainty values:")
print(idhx_air_in_T_deg_F.uncertainty)
```

The function *convert()* does not recalculate the uncertainty based on the original equation, but you can recalculate it even after the unit conversion.
